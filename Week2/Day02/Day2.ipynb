## Day 2 — Probability & Distributions

```python
import numpy as np
import matplotlib.pyplot as plt
```

### task1: Simulate 10,000 coin flips with NumPy and confirm the proportion of heads approaches 0.5.

```python
np.random.seed(42)  # Set a random seed for reproducibility

flips = np.random.randint(0, 2, size=10000) # head = 1, tail = 0

proportion_heads = flips.mean()
print(f"Proportion of heads: {proportion_heads:.4f}")
print(f"Expected: 0.5")
```

**Result:** The proportion of heads came out to 0.4987, very close to the 
theoretical probability of 0.5 (a difference of only 0.0013). This demonstrates 
the **Law of Large Numbers**: as the number of trials increases, the observed 
proportion converges toward the theoretical probability. With only 10 flips, 
the proportion could easily land far from 0.5 due to randomness — but with 
10,000 flips, that randomness "averages out," and the result approaches the 
true value.

### task2: Sample from a normal distribution with np.random.normal and plot its histogram to confirm the bell shape.

```python
normal_sample = np.random.normal(loc=0, scale=1, size=10000)

plt.figure(figsize=(8, 5))
plt.hist(normal_sample, bins=50, edgecolor='black', color='skyblue')
plt.title("Histogram of Normal Distribution Sample")
plt.xlabel("Value")
plt.ylabel("Frequency")
plt.show()

print(f"Sample mean: {normal_sample.mean():.4f}")
print(f"Sample std: {normal_sample.std():.4f}")
```

**Result:** The histogram came out bell-shaped and symmetric around 0, with 
most values clustered near the mean and extreme values becoming rarer the 
farther they are from it. The sample mean and standard deviation came out 
very close to the theoretical values (0 and 1), which confirms that the 
sampling process from `np.random.normal` is working correctly.

### task3: Given a small worked scenario, compute a conditional probability by hand and verify it with a simulation.

```python
np.random.seed(1)

n = 1_000_000
balls = np.random.randint(0, 10, size=n)

is_red = balls < 6
is_win = balls < 3   

# P(Win | Red) = P(Win and Red) / P(Red)
p_win_given_red = (is_win & is_red).sum() / is_red.sum()
print(f"P(Win | Red) simulated: {p_win_given_red:.4f}")
print(f"P(Win | Red) calculated: {0.5}")
```

**Result:** The simulation gave 0.4993, matching the hand calculation of 0.5. 
This illustrates the core idea of conditional probability: once we know a ball 
is red, **the sample space shrinks** — instead of considering all 10 balls, 
we now only consider the 6 red ones, of which 3 are winning. This is exactly 
the same logic used in classification models: instead of asking "what's the 
probability of churn?" in general, we ask "what's the probability of churn 
given that the customer is in a certain age group?" — the condition narrows 
down the possible scenarios.
