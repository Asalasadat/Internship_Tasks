## Day 1 — Descriptive Statistics

### task1: Load a numeric column from a provided dataset into a NumPy array or Pandas Series.

```python
import pandas as pd
import numpy as np
df = pd.read_csv('Mobile Price/train.csv')
df.columns
```

```python
battery_power = df['battery_power']
battery_power # type: pandas.core.series.Series
```

### task2: Compute its mean, median, mode, standard deviation, and IQR.

```python
mean_battery_power = battery_power.mean()
median_battery_power = battery_power.median()
mode_battery_power = battery_power.mode()[0]  # mode() returns a Series, take the first value
std_battery_power = battery_power.std()
IQR_battery_power = battery_power.quantile(0.75) - battery_power.quantile(0.25)
print(f"Mean: {mean_battery_power}")
print(f"Median: {median_battery_power}")
print(f"Mode: {mode_battery_power}")
print(f"Standard Deviation: {std_battery_power}")
print(f"Interquartile Range (IQR): {IQR_battery_power}")
```

Statistical analysis of the battery_power column shows a mean of 1239.06 and a median of 1228.0, and their closeness indicates the data is roughly symmetric, without strong skewness or influential outliers. The mode came out at 618.0, the most frequently occurring value, despite its relative distance from the center — which is normal for continuous, varied data like battery capacity. The standard deviation was 439.29, roughly 35% of the mean, reflecting moderate-to-high spread across devices. The interquartile range (IQR) was 763.75, a large value consistent with the standard deviation, confirming that the spread comes from genuine variation in battery specifications across devices rather than from outliers requiring cleaning. Overall, the column represents a healthy, well-behaved distribution suitable for use as a feature in a subsequent classification model.

### task3: Identify whether the mean or median better represents a 'typical' value, and justify the choice.

The mean and median for battery_power are nearly identical (1239.06 vs. 1228.0, a difference of less than 1%), which indicates the distribution is roughly symmetric with no significant skew or influential outliers. Because of this, the mean is the better representative of a "typical" value: it uses all the data points rather than just the middle position, and it integrates naturally with other statistics like standard deviation for any downstream analysis. The usual reason to prefer the median — protection against distortion by extreme values — doesn't apply here, since the IQR (763.75) is consistent with the standard deviation (439.29), confirming there's no meaningful outlier problem. Had the mean and median diverged substantially, that would have signaled skew, and the median would have become the safer choice instead.
