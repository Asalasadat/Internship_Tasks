## Tokenization

A raw text sample was taken from the **WELFake dataset** and tokenized using NLTK's `word_tokenize()` function.

Tokenization splits the raw text into individual **tokens**, such as words, numbers, and punctuation marks.

The tokenized output will be used as a basis for the subsequent text preprocessing steps.


```python
import re
import string
import nltk

from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
from nltk.tokenize import word_tokenize

# Download required NLTK resources
nltk.download("stopwords")
nltk.download("wordnet")
nltk.download("omw-1.4")

# Initialize tools
stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

# 1. Lowercase
cleaned_text = raw_text.lower()

# 2. Tokenize
tokens = word_tokenize(cleaned_text)

# 3. Remove punctuation
tokens = [
    word for word in tokens
    if word not in string.punctuation
]

# 4. Remove stop words
tokens = [
    word for word in tokens
    if word not in stop_words
]

# 5. Lemmatization
tokens = [
    lemmatizer.lemmatize(word)
    for word in tokens
]

# Combine tokens back into text
cleaned_text = " ".join(tokens)

print("Cleaned text:")
print(cleaned_text)
```

## Full Text Cleaning Pipeline

A complete text preprocessing pipeline was applied to the raw text sample.

The pipeline consists of the following steps:

1. **Lowercasing** — converts all words to lowercase to ensure consistency.
2. **Punctuation Removal** — removes punctuation marks that are not needed for the analysis.
3. **Stop Word Removal** — removes common English words such as `the`, `is`, and `to`.
4. **Lemmatization** — reduces words to their base or dictionary form while preserving their meaning.

The final output is a cleaned text that is ready for further Natural Language Processing (NLP) tasks.


```python
# Verify that important negation words are preserved
negation_words = {
    "no", "not", "nor", "never", "neither", "nobody",
    "nothing", "nowhere", "hardly", "scarcely", "barely"
}

preserved_negations = [
    word for word in tokens
    if word in negation_words
]

print("Preserved negation words:")
print(preserved_negations)

print("\nNumber of preserved negation words:", len(preserved_negations))
```

```python
# Check which negation words are included in NLTK stop words
negation_words = {
    "no", "not", "nor", "never", "neither",
    "nobody", "nothing", "nowhere",
    "hardly", "scarcely", "barely"
}

print("Negation words removed by NLTK:")
print(negation_words.intersection(stop_words))
```

## Verification of Task-Critical Words

The cleaning pipeline was reviewed to ensure that task-critical words were preserved.

The NLTK English stop-word list includes some important negation words, such as `no`, `not`, and `nor`. Removing these words could change the meaning of a statement and potentially affect text classification.

Therefore, these negation words were excluded from the stop-word removal list and preserved during preprocessing.

### Cleaning Choices

* **Lowercasing:** Applied to standardize the text.
* **Punctuation removal:** Applied to remove punctuation that is not essential for the classification task.
* **Stop-word removal:** Applied to reduce common words with limited information.
* **Negation preservation:** `no`, `not`, and `nor` were explicitly preserved because they can affect the meaning of a statement.
* **Lemmatization:** Applied to reduce words to their base forms while retaining their meaning.

The stop-word list was customized to prevent important negation words from being removed.
