## Day 1 — Supervised Learning Concepts & the Scikit-learn API

```python
import pandas as pd
from sklearn.model_selection import train_test_split
```

### task1: Load a provided dataset and separate it into features X and target y.

```python
df = pd.read_csv('Students/StudentsPerformance.csv')
df.head()
```

```python
x = df.drop('math score', axis=1)
y = df['math score']
```

The dataset is divided into **features (X)** and a **target variable (y)**.

- `X` contains all input features that will be used by the model for prediction.
- `y` contains the target variable we want the model to predict.
- Here, **Math Score** is selected as the target variable.

Therefore, `math score` is removed from `X` and stored separately in `y`.

### task2: Perform an 80/20 train/test split with a fixed random_state.

```python
X_train, X_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)
```

The dataset is divided into two subsets:

- **80% of the data** is used for training the model.
- **20% of the data** is reserved for testing the model on unseen data.
- `random_state=42` is used to ensure that the split is reproducible.

###  task3: Confirm the shapes of X_train, X_test, y_train, y_test are consistent and print them.

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```
